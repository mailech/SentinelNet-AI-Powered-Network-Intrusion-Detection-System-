```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

from sklearn.ensemble import RandomForestClassifier, IsolationForest
from sklearn.svm import OneClassSVM
from sklearn.neighbors import LocalOutlierFactor
from sklearn.cluster import KMeans

from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data_path = r"C:\Users\Admin\data\train_processed_advanced.csv"

df = pd.read_csv(data_path)

print("Dataset Shape:", df.shape)
df.head()
```

```python
target_column = df.columns[-1]

X = df.drop(target_column, axis=1)
y = df[target_column]

print("Target column:", target_column)
print("Feature shape:", X.shape)
print("Label shape:", y.shape)
```

```python
cat_cols = X.select_dtypes(include="object").columns

le = LabelEncoder()

for col in cat_cols:
    X[col] = le.fit_transform(X[col])

print("Categorical columns encoded:")
print(cat_cols)
```

```python
label_encoder = LabelEncoder()

y = label_encoder.fit_transform(y)

print("Encoded labels:")
print(np.unique(y))
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Scaled feature matrix shape:", X_scaled.shape)
```

```python
iso = IsolationForest(contamination=0.02)

iso_pred = iso.fit_predict(X_scaled)

print("Isolation Forest Results")
print("Normal:", np.sum(iso_pred == 1))
print("Anomaly:", np.sum(iso_pred == -1))
```

```python
data_path = r"C:\Users\Admin\data\train_processed_advanced.csv"

df = pd.read_csv(data_path)

print("Dataset Shape:", df.shape)

df.head()
```

```python
print(df.columns)
```

```python
X = df.drop("attack_class", axis=1)

y = df["attack_class"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Scaling completed")
print(X_scaled.shape)
```

```python
print(type(X_scaled))
```

```python
iso = IsolationForest(contamination=0.02, random_state=42)

iso_pred = iso.fit_predict(X_scaled)

normal = np.sum(iso_pred == 1)
anomaly = np.sum(iso_pred == -1)

print("Isolation Forest Results")
print("Normal:", normal)
print("Anomaly:", anomaly)

plt.figure(figsize=(6,4))
sns.barplot(x=["Normal","Anomaly"], y=[normal, anomaly])
plt.title("Isolation Forest Anomaly Detection")
plt.ylabel("Count")
plt.show()
```
