```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
import warnings
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, LabelEncoder, OneHotEncoder, PowerTransformer
from sklearn.model_selection import train_test_split, StratifiedKFold
# from sklearn.feature_selection import SelectKBest, chi2, f_classif, RFE, MutualInfoClassifier
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
```

```python
from sklearn.ensemble import RandomForestClassifier, IsolationForest
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer, KNNImputer
```

```python
warnings.filterwarnings('ignore')

# Plot styling
plt.style.use('ggplot')
sns.set_palette("husl")

# Define Column Names for NSL-KDD (Standard List)
COL_NAMES = ["duration","protocol_type","service","flag","src_bytes",
    "dst_bytes","land","wrong_fragment","urgent","hot","num_failed_logins",
    "logged_in","num_compromised","root_shell","su_attempted","num_root",
    "num_file_creations","num_shells","num_access_files","num_outbound_cmds",
    "is_host_login","is_guest_login","count","srv_count","serror_rate",
    "srv_serror_rate","rerror_rate","srv_rerror_rate","same_srv_rate",
    "diff_srv_rate","srv_diff_host_rate","dst_host_count","dst_host_srv_count",
    "dst_host_same_srv_rate","dst_host_diff_srv_rate","dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate","dst_host_serror_rate","dst_host_srv_serror_rate",
    "dst_host_rerror_rate","dst_host_srv_rerror_rate","label", "difficulty_level"]
# Load Data (Assuming data is in ../SentinelNet/data or ../data relative to notebook)
# You might need to adjust the path based on your exact folder structure
train_path = './KDDTest+.txt'
test_path = './KDDTest+.txt'

# Try loading with error handling for paths
try:
    df_train = pd.read_csv(train_path, header=None, names=COL_NAMES)
    df_test = pd.read_csv(test_path, header=None, names=COL_NAMES)
    print(f"Training Data Shape: {df_train.head()}")
    print(f"Test Data Shape: {df_test.tail()}")
except FileNotFoundError:
    print("Files not found. Please check the paths. Trying absolute paths or checking local directory...")
    # Fallback to local directory if paths are different
    if os.path.exists('KDDTrain+.txt'):
        df_train = pd.read_csv('KDDTrain+.txt', header=None, names=COL_NAMES)
        df_test = pd.read_csv('KDDTest+.txt', header=None, names=COL_NAMES)
```

```python
# Correlation Heatmap (only numerical features)

plt.figure(figsize=(14,10))

numeric_df = df_train.select_dtypes(include=['int64','float64'])

corr = numeric_df.corr()

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
# Feature Engineering

df_train['attack_class'] = df_train['label'].apply(
    lambda x: 0 if x == 'normal' else 1
)

df_test['attack_class'] = df_test['label'].apply(
    lambda x: 0 if x == 'normal' else 1
)

print(df_train['attack_class'].value_counts())
```

```python
# Label Encoding

categorical_cols = ['protocol_type','service','flag']

encoder = LabelEncoder()

for col in categorical_cols:
    df_train[col] = encoder.fit_transform(df_train[col])
    df_test[col] = encoder.transform(df_test[col])
```

```python
# Feature Selection

X_train = df_train.drop(['label','attack_class','difficulty_level'], axis=1)
y_train = df_train['attack_class']

X_test = df_test.drop(['label','attack_class','difficulty_level'], axis=1)
y_test = df_test['attack_class']

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
```

```python
#SMOTE
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_resampled, y_resampled = smote.fit_resample(X_train, y_train)

print("Before SMOTE:", y_train.value_counts())
print("After SMOTE:", pd.Series(y_resampled).value_counts())
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=20)

X_resampled_pca = pca.fit_transform(X_resampled)
X_test_pca = pca.transform(X_test)

print("PCA Shape:", X_resampled_pca.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=150,
    max_depth=12,
    random_state=42
)
```

```python
rf_model.fit(X_resampled_pca, y_resampled)

print("Model Training Completed")
```

```python
# Train Random Forest without PCA for importance

rf_feature = RandomForestClassifier(
    n_estimators=150,
    random_state=42
)

rf_feature.fit(X_resampled, y_resampled)

importances = rf_feature.feature_importances_

feature_imp = pd.Series(
    importances,
    index=X_train.columns
).sort_values(ascending=False)

# Plot Top 20

plt.figure(figsize=(10,6))

feature_imp.head(20).plot(kind='barh')

plt.title("Top 20 Important Features")
plt.gca().invert_yaxis()

plt.show()
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = rf_model.predict(X_test_pca)

print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report\n")
print(classification_report(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6,5))

sns.heatmap(cm, annot=True, fmt='d', cmap="Blues")

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.title("Confusion Matrix")

plt.show()
```
