## Imports

```python
import os
import requests
import numpy as np
import pandas as pd

%matplotlib inline
print("Libraries imported")
```

## All Files Path and Locations

```python
# The Path in which Data is Stored
DATA_DIR = "data"
os.makedirs(DATA_DIR, exist_ok=True)

# The URL for Dataset
URLS = {
    "KDDTrain+.txt": "https://raw.githubusercontent.com/defcom17/NSL_KDD/master/KDDTrain+.txt",
    "KDDTest+.txt":  "https://raw.githubusercontent.com/defcom17/NSL_KDD/master/KDDTest+.txt",
}

# All Columns are Labeled
COLUMNS = [
    "duration", "protocol_type", "service", "flag",
    "src_bytes", "dst_bytes", "land", "wrong_fragment", "urgent", "hot",
    "num_failed_logins", "logged_in", "num_compromised", "root_shell",
    "su_attempted", "num_root", "num_file_creations", "num_shells",
    "num_access_files", "num_outbound_cmds", "is_host_login", "is_guest_login",
    "count", "srv_count", "serror_rate", "srv_serror_rate", "rerror_rate",
    "srv_rerror_rate", "same_srv_rate", "diff_srv_rate", "srv_diff_host_rate",
    "dst_host_count", "dst_host_srv_count", "dst_host_same_srv_rate",
    "dst_host_diff_srv_rate", "dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate", "dst_host_serror_rate",
    "dst_host_srv_serror_rate", "dst_host_rerror_rate",
    "dst_host_srv_rerror_rate", "class", "difficulty_level"
]

print(f"Total columns : {len(COLUMNS)}")
```

## Downlaad Data without any Error

```python
def download_file(url, filename):
    path = os.path.join("..", DATA_DIR, filename)
    if os.path.exists(path):
        print(f"{filename} already exists")
        return
    print(f"Downloading {filename}")
    try:
        response = requests.get(url, stream=True)
        response.raise_for_status()
        with open(path, "wb") as f:
            
            # It is chunked to avoid downloading everything as a whole
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)
        print(f"Saved to {path}")
    except Exception as e:
        print(f"Failed to download {filename}: {e}")


print("Downloading dataset files: ")
for filename, url in URLS.items():
    download_file(url, filename)
print("\nAll files should now be in 'data' folder")
```

## Load and Label the Dataset

```python
# Assigning Path for Train and Test Data
train_path = os.path.join(DATA_DIR, "KDDTrain+.txt")
test_path  = os.path.join(DATA_DIR, "KDDTest+.txt")

# Assigning Variables for Train and Test Data
train_df = pd.read_csv(train_path, header=None, names=COLUMNS)
test_df  = pd.read_csv(test_path,  header=None, names=COLUMNS)

print(f"Training set : {train_df.shape[0]:}")
print(f"Test set     : {test_df.shape[0]:}")
```

## 6. Train vs Test Set Comparison

Verify the test set has a representative sample of attack types seen in training.

```python
# Side-by-side class distribution: train vs test
train_dist = train_df['class'].value_counts(normalize=True).rename('Train %')
test_dist  = test_df['class'].value_counts(normalize=True).rename('Test %')
comparison = pd.concat([train_dist, test_dist], axis=1).fillna(0).mul(100).round(2)

print('Class Distribution — Train vs Test (%):')
print(comparison.sort_values('Train %', ascending=False).to_string())
```

## 7. Column Reference

| Column | Description |
|--------|-------------|
| `duration` | Connection length in seconds |
| `protocol_type` | Protocol used: `tcp`, `udp`, `icmp` |
| `service` | Network service at destination (e.g. `http`, `ftp`) |
| `flag` | Connection status (e.g. `SF` = normal, `S0` = no reply) |
| `src_bytes` | Bytes sent from source to destination |
| `dst_bytes` | Bytes sent from destination back to source |
| `land` | 1 if source and destination IP/port are the same (Land DoS) |
| `wrong_fragment` | Number of malformed packet fragments |
| `urgent` | Packets with urgent bit set |
| `hot` | Number of "hot" indicators (e.g. accessing root directories) |
| `num_failed_logins` | Failed login attempts |
| `logged_in` | 1 if user successfully logged in |
| `num_compromised` | Conditions indicating system compromise |
| `root_shell` | 1 if root shell was obtained |
| `su_attempted` | 1 if `su` to root was attempted |
| `num_root` | Root accesses performed |
| `num_file_creations` | File creation operations |
| `num_shells` | Shell prompts generated |
| `num_access_files` | Attempts to access critical files (e.g. `/etc/passwd`) |
| `num_outbound_cmds` | Outbound commands in FTP session |
| `is_host_login` | 1 if login belongs to host admin list |
| `is_guest_login` | 1 if login is a guest login |
| `count` | Connections to same host in past 2 seconds |
| `srv_count` | Connections to same service in past 2 seconds |
| `serror_rate` | % connections with SYN errors (same host) |
| `srv_serror_rate` | % connections with SYN errors (same service) |
| `rerror_rate` | % connections with REJ errors (same host) |
| `srv_rerror_rate` | % connections with REJ errors (same service) |
| `same_srv_rate` | % connections to same host → same service |
| `diff_srv_rate` | % connections to same host → different service |
| `srv_diff_host_rate` | % connections to same service from different hosts |
| `dst_host_count` | Connections to same destination host (last 100) |
| `dst_host_srv_count` | Connections to same host + service (last 100) |
| `dst_host_same_srv_rate` | % connections to same service |
| `dst_host_diff_srv_rate` | % connections to different services |
| `dst_host_same_src_port_rate` | % connections from same source port |
| `dst_host_srv_diff_host_rate` | % connections to same service from different hosts |
| `dst_host_serror_rate` | % SYN errors to this destination host |
| `dst_host_srv_serror_rate` | % SYN errors to this host + service |
| `dst_host_rerror_rate` | % REJ errors to this destination host |
| `dst_host_srv_rerror_rate` | % REJ errors to this host + service |
| `class` | **Target variable** — `normal` or attack name |
| `difficulty_level` | Classification difficulty score (0–21) |
